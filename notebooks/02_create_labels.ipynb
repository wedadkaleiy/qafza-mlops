```python
import pandas as pd

ml_table = pd.read_csv("../processedData/ml_table_raw.csv")
print(ml_table.shape)
```

```python
# تحويل الأعمدة من نص (string) لتاريخ حقيقي (datetime) عشان نقدر نقارنهم
ml_table["order_delivered_customer_date"] = pd.to_datetime(ml_table["order_delivered_customer_date"])
ml_table["order_estimated_delivery_date"] = pd.to_datetime(ml_table["order_estimated_delivery_date"])

# بناء الـ label: 1 لو الطلب متأخر، 0 لو وصل بالوقت أو قبله
ml_table["is_late"] = (ml_table["order_delivered_customer_date"] > ml_table["order_estimated_delivery_date"]).astype(int)

print(ml_table["is_late"].value_counts())
```

### Class Imbalance Check

- Total orders: 99,441
- On-time (0): 91,614 (~92.13%)
- Late (1): 7,827 (~7.87%)

**Conclusion:** Clear class imbalance problem — late orders represent 
only ~7.87% of the data. Accuracy alone would be misleading (a model 
predicting "always on-time" would score ~92% without learning anything).
This will be addressed in Notebook 6 by choosing an appropriate metric 
(e.g., precision/recall, F1, or ROC-AUC) instead of relying on accuracy alone.

```python
check_sample = ml_table[["order_id", "order_delivered_customer_date", 
                           "order_estimated_delivery_date", "is_late"]].sample(5, random_state=42)
print(check_sample)
```
