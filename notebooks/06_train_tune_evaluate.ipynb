```python
import pandas as pd
from sklearn.metrics import classification_report
from sklearn.linear_model import LogisticRegression
```

```python
X_train = pd.read_csv("../processedData/X_train.csv")
y_train = pd.read_csv("../processedData/y_train.csv").squeeze()
X_val = pd.read_csv("../processedData/X_val.csv")
y_val = pd.read_csv("../processedData/y_val.csv").squeeze()
X_test = pd.read_csv("../processedData/X_test.csv")
y_test = pd.read_csv("../processedData/y_test.csv").squeeze()
```

```python
baseline_preds = [0] * len(y_val)
print(classification_report(y_val, baseline_preds))
```

```python
model = LogisticRegression(max_iter=1000, class_weight="balanced")
model.fit(X_train, y_train)

val_preds = model.predict(X_val)
print(classification_report(y_val, val_preds))
```

```python
import joblib

joblib.dump(model, "../processedData/final_model.pkl")

with open("../processedData/results_summary.md", "w") as f:
    f.write("""# Model Results Summary

## Baseline (always predict 0)
Recall (class 1): 0%

## Logistic Regression (class_weight=balanced)
Recall (class 1): 51%
Precision (class 1): 8%
F1 (class 1): 0.14

## Conclusion
Model beats baseline on recall for the minority class (late deliveries),
which is the priority metric given the imbalance. Precision is low —
future improvement: try stronger models (Random Forest, XGBoost) or
tune the classification threshold.
""")

print("Saved!")
```
